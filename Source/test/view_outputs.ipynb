### Output Viewer
#### NOTE
The purpose of this notebook is to view the outputs of the RnR post-processing script. Saved outputs from the mock-data will be used, with a provided code snippet for viewing outputs from the `docker/run_post_process.sh`

```python
from pathlib import Path

import geopandas as gpd
from shapely import wkb

output_data_dir = Path("../data/output/")

# The file name should look something like: output_inundation_2025-06-06_14:28:32.csv
file_name = "output_inundation_2025-09-25_15:11:53.csv"
```

#### Data Viewer

If you want to do the same mapping, but for the outputs from the RnR containers, run the following (NOTE: We're changing wkb -> wkt):

```python
import geopandas as gpd
from shapely import wkb
import ast

def convert_geom(x):
    if x is None:
        return None
    elif isinstance(x, str) and x.startswith("b'"):
        try:
            # Convert string representation of bytes back to actual bytes
            byte_data = ast.literal_eval(x)
            geom = wkb.loads(byte_data)
            return geom
        except Exception as e:
            print(f"Error converting geometry: {e}")
            return None
    else:
        print(f"Unexpected type: {type(x)}")
        return x

df = gpd.read_file(output_data_dir / file_name)
```

```python
df["geom"] = df["geom"].apply(convert_geom)
df_clean = df[df["geom"].notna() & df["geom"].apply(lambda x: hasattr(x, 'geom_type'))]

gdf = gpd.GeoDataFrame(df_clean.drop(["geom"], axis=1), geometry=df_clean["geom"], crs="EPSG:5070")
gdf = gdf.to_crs(epsg=3857)
gdf.head()
```
