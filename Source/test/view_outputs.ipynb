### Output Viewer
#### NOTE
The purpose of this notebook is to view the outputs of the RnR post-processing script. Saved outputs from the mock-data will be used, with a provided code snippet for viewing outputs from the `docker/run_post_process.sh`

```python
from pathlib import Path

import geopandas as gpd
from shapely import wkt

output_data_dir = Path("../data/output/")

# The file name should look something like: output_inundation_2025-06-06_14:28:32.csv
file_name = "output_inundation_2025-06-06_14:28:32.csv"
```

#### Data Viewer

If you want to do the same mapping, but for the outputs from the RnR containers, run the following:

```python
df = gpd.read_file(output_data_dir / file_name)
df["geom"] = df["geom"].apply(lambda x: wkt.loads(x) if isinstance(x, str) else x)
gdf = gpd.GeoDataFrame(df.drop(["geom"], axis=1), geometry=df["geom"], crs="EPSG:5070")
gdf = gdf.to_crs(epsg=3857)
gdf.head()
```

```python
gdf.explore()
```
