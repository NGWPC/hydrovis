### Output Viewer
The purpose of this notebook is to view the outputs of the RnR post-processing script. Saved outputs from the mock-data will be used, with a provided code snippet for viewing outputs from the `docker/run_post_process.sh`

```python
from pathlib import Path

import geopandas as gpd
from shapely import wkt

test_data_dir = Path("./mock_data/")
output_data_dir = Path("../data/output/")
```

##### Mock Data Viewer

The snippet below will assist in viewing the flowlines and internal data generated from Replace and Route

```python
df = gpd.read_file(test_data_dir / "output_inundation_20250519_142103.csv")
df["geometry"] = df["geometry"].apply(
    lambda x: wkt.loads(x) if isinstance(x, str) else x
)
gdf = gpd.GeoDataFrame(
    df.drop(["geometry"], axis=1), geometry=df["geometry"], crs="EPSG:5070"
)
gdf = gdf.to_crs(epsg=3857)
```
